```python
import numpy as np
import matplotlib.pyplot as plt

v0 = 10           # kecepatan awal (m/s)
mu_k = 0.20       # koefisien gesek kinetik
g = 9.81          # percepatan gravitasi (m/s^2)

# perhitungan

# perlambatan akibat gesekan
a = -mu_k * g

# waktu sampai benda berhenti
t_stop = v0 / (mu_k * g)

print("Percepatan akibat gesekan =", a, "m/s^2")
print("Waktu benda berhenti       =", t_stop, "s")

# DATA WAKTU


t = np.linspace(0, t_stop, 100)

 # Persamaan kecepatan
v = v0 + a * t

 # agar kecepatan tidak menjadi negatif
v[v < 0] = 0

 # GRAFIK

plt.figure(figsize=(10, 6))

plt.plot(t, v, linewidth=2, label="v(t)")

plt.scatter(t_stop, 0, s=60,
            label=f"Benda berhenti (t = {t_stop:.2f} s)")

plt.xlabel("waktu, t (s)")
plt.ylabel("Kecepatan, v (m/s)")
plt.title("Grafik Kecepatan Benda Akibat Gesekan")

plt.grid(True)
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Parameter Fisika dan Simulasi ---
g = 9.8         # Percepatan gravitasi (m/s^2)
gamma = 1.0     # Koefisien hambatan udara (1/s)
dt = 0.01       # Langkah waktu (delta t)
t_max = 5.0     # Total waktu simulasi (detik)

# --- Inisialisasi Array ---
t = np.arange(0, t_max + dt, dt)
v_num = np.zeros(len(t))  # Array untuk solusi numerik (Euler)
v_ana = np.zeros(len(t))  # Array untuk solusi analitis

# Kondisi awal: kecepatan mula-mula v(0) = 0
v_num[0] = 0.0

# --- Loop Metode Euler ---
for i in range(len(t) - 1):
    # Persamaan diferensial: dv/dt = g - gamma * v
    dv_dt = g - gamma * v_num[i]
    # Integrasi Euler: v(t + dt) = v(t) + (dv/dt) * dt
    v_num[i+1] = v_num[i] + dv_dt * dt

# --- Solusi Analitis ---
# v(t) = (g / gamma) * (1 - e^(-gamma * t))
v_ana = (g / gamma) * (1 - np.exp(-gamma * t))

# --- Plot Hasil Perbandingan ---
plt.figure(figsize=(8, 5))
plt.plot(t, v_num, 'r--', label=f'Numerik Euler ($\Delta t$ = {dt})', linewidth=2)
plt.plot(t, v_ana, 'b-', label='Solusi Analitis', alpha=0.7)
plt.title('Perbandingan Kecepatan v(t) Numerik vs Analitis')
plt.xlabel('Waktu t (sekon)')
plt.ylabel('Kecepatan v (m/s)')
plt.grid(True)
plt.legend()
plt.show()
```
